```python
# Crop the central 200×200 pixel area using Matplotlib
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

h, w = img.shape[:2]

cropped = img[h//2-100:h//2+100,
              w//2-100:w//2+100]

plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
# Convert RGB to HSV and display H, S and V channels separately
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)

h, s, v = cv2.split(hsv)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(h, cmap="gray")
plt.title("Hue")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(s, cmap="gray")
plt.title("Saturation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(v, cmap="gray")
plt.title("Value")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Calculate Mean Squared Error (MSE) between two grayscale images
```

```python
import cv2
import numpy as np

image = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
images = cv2.imread("images.jpg", cv2.IMREAD_GRAYSCALE)

if image is None or images is None:
    raise FileNotFoundError("Image file not found")

if image.shape != images.shape:
    raise ValueError("Both images must have the same dimensions")

mse = np.mean((image.astype(float) - images(2).astype(float)) ** 2)

print("Mean Squared Error:", mse)
```
